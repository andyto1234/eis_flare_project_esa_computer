```python
import pandas as pd
import glob
import os
```

```python

# Define the path pattern for CSV files
path_pattern = '/Users/andysh.to/Script/Python_Script/eis_flare_project_esa_computer/flare_dfs/filtered_flares_2*.csv'

# Get list of all matching CSV files
csv_files = glob.glob(path_pattern)

# Read and concatenate all CSV files into one dataframe
dfs = []
for file in csv_files:
    df = pd.read_csv(file)
    dfs.append(df)
    
# Combine all dataframes
combined_df = pd.concat(dfs, ignore_index=True)

# Optional: Sort by date if there's a date column
# combined_df = combined_df.sort_values('date_column_name')
```

```python
combined_df['stud_acr']
```

```python
# Filter the dataframe to only include rows where 'stud_acr' starts with 'Altas'
filtered_df = combined_df[combined_df['stud_acr'].str.startswith('Atlas', na=False)]
filtered_df
```

```python

# Define the path pattern for CSV files
path_pattern = '/Users/andysh.to/Script/Python_Script/eis_flare_project_esa_computer/flare_dfs/filtered_flares_by_date_*.csv'

# Get list of all matching CSV files
flare_csv_files = glob.glob(path_pattern)

# Read and concatenate all CSV files into one dataframe
dfs = []
for file in flare_csv_files:
    df = pd.read_csv(file)
    dfs.append(df)
    
# Combine all dataframes
flare_df = pd.concat(dfs, ignore_index=True)

# Optional: Sort by date if there's a date column
# combined_df = combined_df.sort_values('date_column_name')
```

```python
# Merge filtered_df with flare_df based on matching times
# First rename the column in flare_df to match filtered_df for merging
flare_df_renamed = flare_df.rename(columns={'EIS flare times': 'flare times'})

# Merge the dataframes on 'flare times' column
filtered_df = pd.merge(filtered_df, 
                      flare_df_renamed[['flare times', 'Start Time', 'Peak Time', 'End Time']], 
                      on='flare times',
                      how='left')

# Display the result
# Rename 'flare times' to 'eis filename time' and display the result
filtered_df = filtered_df.rename(columns={'flare times': 'eis filename time'})
filtered_df
```

```python
# Export the filtered dataframe to CSV
filtered_df.to_csv('filtered_dfs/atlas_flares.csv', index=False)
```
